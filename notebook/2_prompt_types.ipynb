Dynamic Prompts

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    {"role" : "system", "content": "You are a language translator, translate the text into{language}"},
    {"role": "user","content":"{query}"}
])
```

```python
final_prompt = prompt.format_messages(language = "kannada", query = "Hi how are you?")
```

```python
#res = llm.invoke(final_prompt)
#res.content
```

```python
prompt.invoke({"language": "kannada", "query": "Hi how are you?"})
```

```python
res = llm.invoke(prompt.invoke({"language": "kannada", "query": "Hi how are you?"}))
res.content
```

```python
chain = prompt | llm
```

```python
res = chain.invoke({"language": "Hinglish", "query": "Hi how are you?"})
res.content
```

```python
res
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
out =StrOutputParser()
```

```python
out.invoke(res)
```

```python
chain = prompt | llm | out
```

```python
res=chain.invoke({"language":"Hinglish","query":"Hi how are you?"})
res
```

```python
 #Custom function
def typeconversion(res:str):
    return res.upper()
```

```python
chain = prompt | llm | out | typeconversion
```

```python
res = chain.invoke({"language": "hinglish", "query": "Hi how are you?"})
res
```

```python
#custon function
def typeconversion(res:str):
    return res.lower()
```

```python
chain = prompt | llm | out | typeconversion
```
